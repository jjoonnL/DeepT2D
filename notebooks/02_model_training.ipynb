# Full-cohort model training
Train the repeated full-cohort models used for stable subtype labeling. Held-out generalization is evaluated separately in Notebook 03.

```python
from pathlib import Path
import sys
import torch

PROJECT_DIR = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(PROJECT_DIR))
from src import config
from src.full_cohort import train_runs

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
RESULT_DIR = config.OUTPUT_DIR / 'full_cohort'
RUN_FULL_TRAINING = True
print('Device:', DEVICE)
print('Runs:', config.NUM_RUNS)
```

```python
processed = torch.load(config.PROCESSED_DATA_PATH, map_location='cpu', weights_only=False)
print('Participants:', len(processed['participant_manifest']))
```

```python
if RUN_FULL_TRAINING:
    model_manifest = train_runs(processed, config, DEVICE, RESULT_DIR)
    display(model_manifest.tail())
else:
    print('Full-cohort training disabled. Set RUN_FULL_TRAINING = True to train all runs.')
```

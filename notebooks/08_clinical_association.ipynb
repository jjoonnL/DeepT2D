# Downstream clinical association
Included-versus-excluded comparison and adjusted complication associations. The covariate coding below is the revision working specification and must be confirmed before manuscript-final use.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_DIR = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(PROJECT_DIR))
from src import config
from src.clinical import adjusted_complication_models, compare_included_excluded

RUN_INCLUDED_EXCLUDED = False
RUN_COMPLICATION_MODELS = False
INCLUSION_INPUT = config.RAW_DATA_DIR / 'cohort_inclusion_data.csv'
COMPLICATION_INPUT = config.RAW_DATA_DIR / 'complication_model_ready.csv'
RESULT_DIR = config.OUTPUT_DIR / 'clinical_association'
```

```python
if RUN_INCLUDED_EXCLUDED:
    cohort = pd.read_csv(INCLUSION_INPUT)
    comparison = compare_included_excluded(
        cohort, 'included',
        continuous=['age','bmi','hba1c','age_at_diagnosis','diabetes_duration','egfr'],
        categorical=['male','current_smoking','insulin_treatment']
    )
    RESULT_DIR.mkdir(parents=True, exist_ok=True)
    comparison.to_csv(RESULT_DIR / 'included_excluded_comparison.csv', index=False)
    display(comparison)
else:
    print('Included-versus-excluded analysis disabled.')
```

```python
if RUN_COMPLICATION_MODELS:
    complication_data = pd.read_csv(COMPLICATION_INPUT)
    global_tests, odds_ratios, diagnostics = adjusted_complication_models(
        complication_data, RESULT_DIR / 'adjusted_complications'
    )
    display(global_tests)
    display(diagnostics)
else:
    print('Adjusted complication analysis disabled pending final covariate confirmation.')
```
